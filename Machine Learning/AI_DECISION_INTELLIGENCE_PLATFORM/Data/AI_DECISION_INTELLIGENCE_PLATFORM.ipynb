```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)

teams = [f"Team_{i}" for i in range(1, 9)]
sprints = range(1, 21)
```

```python
rows = []
```

```python
for team in teams:
    high_load_streak = 0

    for sprint in sprints:
        allocated = np.random.randint(320, 400)
        actual = allocated + np.random.randint(-40, 120)
        overtime = max(0, actual - allocated)
        utilization = round(actual / allocated, 2)

        tasks_planned = np.random.randint(40, 60)
        tasks_completed = tasks_planned - np.random.randint(0, 10)
        spillover = tasks_planned - tasks_completed
        velocity_dev = round((tasks_completed - tasks_planned) / tasks_planned, 2)

        workload_volatility = round(np.random.uniform(0.1, 0.6), 2)

        if utilization > 1.1:
            high_load_streak += 1
        else:
            high_load_streak = 0

        burnout_risk = 1 if (high_load_streak >= 3 and overtime > 30) else 0
        delivery_risk = 1 if (spillover > 5 and abs(velocity_dev) > 0.2) else 0

        rows.append([
            team,
            sprint,
            allocated,
            actual,
            overtime,
            utilization,
            tasks_planned,
            tasks_completed,
            spillover,
            velocity_dev,
            workload_volatility,
            high_load_streak,
            delivery_risk,
            burnout_risk
        ])
```

```python
columns = [
    "team_id",
    "sprint_id",
    "allocated_hours",
    "actual_hours",
    "overtime_hours",
    "utilization_ratio",
    "tasks_planned",
    "tasks_completed",
    "spillover_tasks",
    "velocity_deviation_pct",
    "workload_volatility",
    "consecutive_high_load_sprints",
    "delivery_risk_label",
    "burnout_risk_label"
]
```

```python
df = pd.DataFrame(rows, columns=columns)
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
```

```python
df = pd.read_csv("ai_decision_intelligence_data.csv")
```

```python
import random

def add_label_noise(labels, noise_ratio=0.08):
    noisy_labels = labels.copy()
    n_noisy = int(len(labels) * noise_ratio)
    noisy_indices = random.sample(range(len(labels)), n_noisy)

    for idx in noisy_indices:
        noisy_labels.iloc[idx] = 1 - noisy_labels.iloc[idx]

    return noisy_labels

df["delivery_risk_label"] = add_label_noise(df["delivery_risk_label"], noise_ratio=0.08)
```

```python
df["burnout_risk_label"] = add_label_noise(
    df["burnout_risk_label"],
    noise_ratio=0.05   # 5% noise only
)
```

```python
delivery_features = [
    "utilization_ratio",
    "spillover_tasks",
    "velocity_deviation_pct",
    "workload_volatility"
]

X_delivery = df[delivery_features]
y_delivery = df["delivery_risk_label"]
```

```python
X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(
    X_delivery, y_delivery, test_size=0.2, random_state=42
)
```

```python
scaler_d = StandardScaler()

X_train_d = scaler_d.fit_transform(X_train_d)
X_test_d = scaler_d.transform(X_test_d)
```

```python
delivery_model = Sequential([
    Dense(8, activation="relu", input_shape=(X_train_d.shape[1],)),
    Dense(4, activation="relu"),
    Dense(1, activation="sigmoid")
])

delivery_model.compile(
    optimizer=Adam(learning_rate=0.01),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
delivery_model.fit(
    X_train_d,
    y_train_d,
    epochs=30,
    batch_size=8,
    validation_split=0.2,
    verbose=0
)
```

```python
delivery_loss, delivery_acc = delivery_model.evaluate(X_test_d, y_test_d, verbose=0)
delivery_acc
```

```python
burnout_features = [
    "utilization_ratio",
    "overtime_hours",
    "consecutive_high_load_sprints",
    "workload_volatility"
]

X_burnout = df[burnout_features]
y_burnout = df["burnout_risk_label"]
```

```python
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_burnout, y_burnout, test_size=0.2, random_state=42
)
```

```python
scaler_b = StandardScaler()

X_train_b = scaler_b.fit_transform(X_train_b)
X_test_b = scaler_b.transform(X_test_b)
```

```python
burnout_model = Sequential([
    Dense(8, activation="relu", input_shape=(X_train_b.shape[1],)),
    Dense(4, activation="relu"),
    Dense(1, activation="sigmoid")
])

burnout_model.compile(
    optimizer=Adam(learning_rate=0.01),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
burnout_model.fit(
    X_train_b,
    y_train_b,
    epochs=30,
    batch_size=8,
    validation_split=0.2,
    verbose=0
)
```

```python
burnout_loss, burnout_acc = burnout_model.evaluate(X_test_b, y_test_b, verbose=0)
burnout_acc
```

```python
delivery_risk_scores = delivery_model.predict(X_test_d).flatten()
```

```python
burnout_risk_scores = burnout_model.predict(X_test_b).flatten()
```

```python
decision_df = df.iloc[X_test_d.shape[0] * -1 :].copy()

decision_df = decision_df.reset_index(drop=True)
decision_df["delivery_risk_score"] = delivery_risk_scores
decision_df["burnout_risk_score"] = burnout_risk_scores
```

```python
def risk_level(score):
    if score >= 0.7:
        return "High"
    elif score >= 0.4:
        return "Medium"
    else:
        return "Low"

decision_df["delivery_risk_level"] = decision_df["delivery_risk_score"].apply(risk_level)
decision_df["burnout_risk_level"] = decision_df["burnout_risk_score"].apply(risk_level)
```

```python
def recommend_action(delivery_level, burnout_level):
    if delivery_level == "High" and burnout_level == "High":
        return "Rebalance workload and escalate risk"
    elif delivery_level == "High":
        return "Add delivery buffer and monitor closely"
    elif burnout_level == "High":
        return "Reduce load and initiate burnout prevention"
    elif delivery_level == "Medium" or burnout_level == "Medium":
        return "Monitor and review weekly"
    else:
        return "No action required"

decision_df["recommended_action"] = decision_df.apply(
    lambda row: recommend_action(
        row["delivery_risk_level"],
        row["burnout_risk_level"]
    ),
    axis=1
)
```

```python
decision_df["decision_confidence"] = decision_df[
    ["delivery_risk_score", "burnout_risk_score"]
].max(axis=1).round(2)
```

```python
final_decision_view = decision_df[
    [
        "team_id",
        "sprint_id",
        "delivery_risk_score",
        "burnout_risk_score",
        "delivery_risk_level",
        "burnout_risk_level",
        "recommended_action",
        "decision_confidence"
    ]
]

final_decision_view.head()
```
